```python
import pandas as pd
```

```python
from sqlalchemy import create_engine    
```

```python
user ='postgres'
password = '123'
host = '10.0.222.18'
port = '5432'
database = 'tallerdb'
```

```python
engine=create_engine(f"postgresql+psycopg2://{user}:{password}@{host}:{port}/{database}")
```

```python
sql = 'select * from familia'
familias = pd.read_sql(sql, engine)
print(familias)

sql = 'select * from categoria'
categorias = pd.read_sql(sql, engine)
print(categorias)
```

```python
vista_cat = categorias.merge(familias, how='inner' , on='idfamilia')
vista_cat



```

```python
vista_cat = categorias.merge (familias, how= 'outer', on='idfamilia')
vista_cat = vista_cat.rename (columns={'nombre_x': 'categoria', 'nombre_y': 'familia'}) 
vista_cat
```

```python
vista_cat_resumen = vista_cat[['categoria', 'familia']]
vista_cat_resumen
```

```python
vista_cat_resumen.groupby('familia')['categoria'].count()
```

```python
vista_cat_resumen.pivot_table(index='familia', columns='categoria', aggfunc='size', fill_value = 0)
```
